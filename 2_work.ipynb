```python
import pandas as pd
```

```python
raw = pd.read_csv(r"C:\Users\rahma khawaja\Downloads\teaching_prices_clean.csv")
```

```python
raw.groupby("item")["price"].agg(["mean"])
```

```python
raw.groupby("item")["price"].agg(["mean", "median"])
```

```python
raw["date"] = pd.to_datetime(raw["date"])

date_prices = raw.groupby(
    ["item", "date"]
)["price"].mean().reset_index()

date_prices = date_prices.sort_values("date")

first_last = date_prices.groupby("item").agg(
    first_price=("price", "first"),
    last_price=("price", "last")
)

first_last["percentage_change"] = (
    (first_last["last_price"] - first_last["first_price"])
    / first_last["first_price"]
) * 100

first_last.round(2)
```
